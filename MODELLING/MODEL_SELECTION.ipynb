**MACHINE LEARNING MODEL**

  - A Machine Learning Model is a Function That tries To Find The Relationship Between the Features & The Target Variables
   
  - It Tries to Find The Pattern in Data, Understand the Data & Trains on the data .

  - Based on this Learning, A Machine Learning Model Makes Predictions & Recognize Patterns

**MODEL SELECTION**
  - Model Selection in Machine Learning is The Process Of Choosing The Best Suited Model For a Particular Problem .

  - Selecting A Model Depends on various Factors Such As The Dataset, Task , Nature  of the Model, etc.


MODELS CAN BE SELECTED BASED ON:

  1. Type of Data Avialable:

     -  Images & Videos    - CNN
     -  Text / Speech Data - RNN
     -  Numerical Data     - SVM, etc

  2. Based On The Task We Need To Carry Out:

     - Classification Task : SVM, Logistic Regression, Decision Tree, etc.

     - Regression Task : Linear Regression, Random Forest, Polynomial regression, etc.

     - Clustering Task : K-Means Clustering, Hierarchial Clustering

**OVERFITTING**
  - Overfitting is a machine learning error where a model learns training data too well, capturing noise and specific fluctuations instead of general patterns.

  - This leads to high accuracy on training data but poor performance on new, unseen data.

CAUSES OF OVERFITTING:
  1. Less Data
  2. Increased Complexity of the Model
  3. More number of Layers in Neural Network  

**UNDERFITTING**
  - Underfitting occurs when a machine learning model is too simple to capture the underlying structure of data, resulting in low accuracy and high error rates on both training and unseen test data.

  -It represents high bias and low variance, indicating the model failed to learn the signal in the data

CAUSES OF UNDERFITTING :

   1. Choosing Wrong Model
   2. Model is Too Simple
   3. Less Variance But High Bias

PREVENT UNDERFITTING BY:

   1. Choosing The Correct Model
   2. Increasing Complexity
   3. Bias-Variance Tradeoff


**BIAS** - Bias is the Difference Between The Average Prediction Of Our Model & The Correct Value Which We are Trying To predict.

**VARIANCE** - Variance is the Amount That The Estimate Of Target Function Will Change if Different Training Data was Used

**BIAS - VARIANCE TRADEOFF**

  - The bias-variance tradeoff is the fundamental balancing act in machine learning between minimizing errors from overly simplistic assumptions [ high bias/underfitting ] and overly complex modeling [ high variance/overfitting ].

  - To Trade-off the Value of Bias & Variance To Get An Optimized Value

Techniques to Have Better BIAS - VARIANCE TRADEOFF :  
   1. Good Model Selection
   2. Regularization
   3. Dimensionality Reductiion
   4. Ensemble Methods

**LOSS FUNCTION**
  - Lopss function Measures How far an Estimated Value is From its True Value

  - It is Helpful To Determine Which Model Performs Better & Which Parameters are better

TYPES OF LOSS FUNCTION :
  1. Cross Entropy Loss
  2. Squared Error Loss
  3. KL Divergence

LOSS FUNCTION = 1/n (y-y)2

Low Loss Value -> High Accuracy

**MODEL EVALUATION**

Regression Evaluation Metrics
Evaluation in machine learning measures the performance of a model.

 Here are some popular evaluation metrics for regression:

**Mean Absolute Error (MAE):** The average absolute difference between the predicted and actual values of the target variable.

**Mean Squared Error (MSE):** The average squared difference between the predicted and actual values of the target variable.

**Root Mean Squared Error (RMSE):** Square root of the mean squared error.

**Huber Loss:** A hybrid loss function that transitions from MAE to MSE for larger errors, providing balance between robustness and MSE’s sensitivity to outliers.

**R2 – Score:** Higher values indicate better fit ranging from 0 to 1.

**TYPES OF PARAMETER**

MODEL PARAMETERS / INTERNEL PARAMETERS
  -  These are variables that the model learns automatically from the training data.
  
  - They are internal to the model and their values are updated during training to minimize error

  - Weights, Biases, Coefficients,Support Vectors, Cluster Centroids

HYPERPARAMETERS / EXTERNAL PARMETERS

  - They are external Parameters whose values control the overall learning process
  
  - These are Adjustable Parameters Used to obtain an optimal Model
   
  - Learning Rate,

GRADIENT DESCENT

   - Optimization refers to Determining best Parameters for a model, Such That the Loss Function of the model Decreases, as a Result of Which The Model Can Predict More Accurately.
