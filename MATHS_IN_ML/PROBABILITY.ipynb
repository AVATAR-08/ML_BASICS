PROBABILITY
 - Probability is a branch of Mathematics that deals with Calculating The  Likelihood Of a Given event to occur.


RANDOM VARIABLES
  - A Random Variable is a Numerical Description Of the Outcomes of Random Events.

TYPES OF RANDOM VARIABLES

  1. Discrete Random Variables
     - A Discrete random Variables Takes Only Discrete / Distinct Values.
     - Ex : Coin Toss

  2. Continious Random Variables
     - A continious Random Variable Can Take Any Value in A Given Range.
     - EX : Weight of Random Person In Class

PROBABILITY DISTRIBUTION FOR RANDOM VARIABLES

  - Probability Distribution For A Random Variable Describes How the Probabilities are Distributed Over The Values Of the Random Variable

NORMAL DISTRIBUTION

  - A Normal Distribution is an Arrangement of A Dataset in Which Most of the Data Points Lie in The middle Of The Range and The Rest Taper off Symmetrically Towards either Extreme

POISSON DISTRIBUTION

  - Poisson Distribution is a Probability Distribution That Measures How Many Times an Events is Likely To Occurs Within a Specified Period Of Time.
  - Poisson Distribution is Used to Understand Independent events that Occur At a Constant Rate Within a Given interval of Time.  



SKEWNESS
 - A Data is Considered Skewed When The Distribution Curve Appears Distorted Curve Appears Distorted / Skewed Either to the Left / to The Right, in a Statistical Distribution.
