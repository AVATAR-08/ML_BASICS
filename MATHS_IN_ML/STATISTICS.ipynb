# **STATISTICS**
- Statistics is the science concerned with developing & studying methods fo collecting, analysing, interpreting & presenting data.


**TYPES OF DATA:**

- CATEGORICAL
    - Nominal Data : A Classification of Categorial Variables, that do not Provide any quantum Value.

    - Ordinal Data : A Type of Data in which the Values Follow a Natural Order.

- NUMERICAL  
    - Discrete : Type of Data that Can only take Certain Values.
    
    - Continious : Continious Data Can Have Almost Any Numerical Value. Unlike Discrete Data,They can Have Decimal Values.


TYPES OF STATISTICS

A. DESCRIPTIVE STATISTICS

  - Describe the Basic Featurs of data
  - Provide Simple Summaries About The Sample & the Measures

  2 important Measures of Descriptive Statistics ::
  
   1. Measures Of Central Tendencies( Mean, Median, Mode)

   2. Measure of Variability(Range, S.D, Variance)

B. INFERENTIAL STATISTICS

  - Takes Data Samples & Makes inferences & Prediction about The Larger Population From Which the Sample Was Drawn

TYPES OF STATISTICAL STUDY

1. Sample Study : It is a Study Which is Carried out on a Sample Which Represents Total Population

2. Observational Study : It is a Study where we simply collect & analyze data. we observe the correlation in data without making any changes.

3. Experimental Study : It is a Study in which Conditions Are Controlled & Manipulated by Experimenter.



POPULATION & SAMPLE

TYPES OF SAMPLING TECHNIQUES:

  1. Simple andom Sampling
        - Samples are picked randomly from a larger population
        - All Datapoints Has an Equal Probability To be Selected

  2. Systematic Sampling
        - Samples is Picked At Regular intervals.
        - Data must Be Evenly Distributed

  3. StratifiedRandom Sampling
        - Population is Divided into Smaller Groups Called Strata.
        - Finds important Characterstics in the Population

  4. Cluster Sampling
        - Population is Subdivided into Clusters & then Random Clusters Are taken as Sample.
        - Reduced Varability
        - Advantages Of Both Random & Sratified Sampling

**CENTRAL TENDENCY**

  - A measure of Central Tendency is a value That represents the Center point / typical Value Of Dataset.

  - it is a value that summarizes the data

**MEASURE OF VARIABILITY**

1. RANGE

    - Range of a set of data is the Difference Between The Largest & Smallest values.
    - it gives rough ideas about distribution of dataset

2. VARIANCE
    
    - Variance is a measure of how for each number in the set is from mean &  therefore from every other number in the dataset.  

3. STANDARD DEVIATION

    - S.D is the square root of variance
    - S.D looks How Spraed out a group of numbers is from the mean

PERCENTILES

  - Percentile is a Value on a Scale of 100 that Indicates the Percent of a Distribution That is Equal or Below it.    
    
  
QUANTILES

  - Quantile is a Measure That Tells How Many Values in A Dataset are Above or Below a Certain Limit.
      
  - It Divides The Members of The Dataset into Equally-sized Subgroups

**CORRELATION**
    
  - Correlation is a Measure That Determines the Extent To Which Two Variables are Related to Each Other in A Dataset.
     
  - But it Doesn't Mean that One Event is The Cause of The Other Event.

POSITIVE CORRELATION

  - One Variable Increases and Another Variables Also Increases
  
NEGATIVE CORRELATION

  - One Variable Increases and Other variable Decreases and Vice-Versa.

CAUSATION

  - Causation Means That One Event Causes Another Event To Occur

  - There is A Cause & Effect RelationShip Between The Two Variables in A dataset


**HYPOTHESIS**

  - Hypothesis is an Assumption That Is Made Based on the Observations of an Experiment

NULL HYPOTHESIS

  - Null Hypothesis(H0) is the Commonly Accepted Fact

ALTERNATIVE HYPOTHESIS

  - Alternative Hypothesis(Ha) is Opposite to Null Hypothesis
  - It Challanges The Null Hypothesis

HYPOTHESIS TESTING

  - Hypothesis is an Assumption That is Made Based On The Observations of an Experiment.
  - Hypothesis Testing is A Method Carried Out to Tests The Assumptions Made in the Experiment
