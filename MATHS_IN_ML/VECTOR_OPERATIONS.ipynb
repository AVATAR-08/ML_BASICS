```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set()
```

PLOTTING A VECTOR

```python
plt.quiver(0,0,2,3)
plt.show()
```

**plt.quiver(X, Y, U, V, [C])**

**X, Y :** The (x, y) coordinates of the arrow locations (starting points).

**U, V :** The (x, y) components of the vector, defining its direction and magnitude.

**C :** (Optional) Array for color mapping, allowing arrow colors to represent a third dimension (e.g., magnitude).

PLOTTING MULTIPLE VECTORS

```python
plt.quiver(0,0,2,3, angles='xy', scale_units='xy', scale=1, color='red')
plt.quiver(0,0,3,2, angles='xy', scale_units='xy', scale=1, color='blue')
plt.xlim(-2,8)     # set the limit of range of x on graph
plt.ylim(-2,8)     # set the limit of range of y on graph
plt.show()
```

ADDITION OF VECTORS

```python
Vector_1=np.asarray([0,0,2,3])
Vector_2=np.asarray([0,0,3,-2])
Add=Vector_1+Vector_2
print(Add)
```

```python
plt.quiver(0,0,2,3, angles='xy', scale_units='xy', scale=1, color='red')
plt.quiver(0,0,3,-2, angles='xy', scale_units='xy', scale=1, color='blue')
plt.quiver(0,0,5,1, angles='xy', scale_units='xy', scale=1, color='y')
plt.xlim(-2,8)     # set the limit of range of x on graph
plt.ylim(-2,8)     # set the limit of range of y on graph
plt.show()
```

SUBTRACTION OF VECTORS

```python
Vector_1=np.asarray([0,0,2,3])
Vector_2=np.asarray([0,0,3,-2])
sub=Vector_1-Vector_2
print(sub)
```

```python
plt.quiver(0,0,2,3, angles='xy', scale_units='xy', scale=1, color='red')
plt.quiver(0,0,3,-2, angles='xy', scale_units='xy', scale=1, color='blue')
plt.quiver(0,0,-1,5, angles='xy', scale_units='xy', scale=1, color='y')
plt.xlim(-8,8)     # set the limit of range of x on graph
plt.ylim(-8,8)     # set the limit of range of y on graph
plt.show()
```

MULTIPLICATION BY A SCALAR

```python
Vector_2=np.asarray([0,0,3,-2])
product=Vector_2*2
print(product)
```

```python
plt.quiver(0,0,3,-2, angles='xy', scale_units='xy', scale=1, color='red')
plt.quiver(0,0,6,-4, angles='xy', scale_units='xy', scale=1, color='blue')
plt.xlim(-8,8)     # set the limit of range of x on graph
plt.ylim(-8,8)     # set the limit of range of y on graph
plt.show()
```

DOT PRODUCT OF TWO VECTORS

```python
A=np.asarray([2,3])
B=np.asarray([4,4])
A_dot_B=np.dot(A,B)
print(A_dot_B)
```

```python
C=np.asarray([14,25,65])
D=np.asarray([44,36,85])
C_dot_D=np.dot(C,D)
print(C_dot_D)
```


CROSS PRODUCT OF TWO VECTORS

```python
A=np.asarray([2,3])
B=np.asarray([4,4])
A_cross_B=np.cross(A,B)
print(A_cross_B)
```

```python
C=np.asarray([14,25,65])
D=np.asarray([44,36,85])
C_cross_D=np.cross(C,D)
print(C_cross_D)
```

PROJECTION OF TWO VECTOR

-projection of A on B = A.B/||B||**2

```python
A=np.asarray([2,3])
B=np.asarray([4,4])
# MAGNITUDE OF "B" VECTOR
Magnitude_of_B= np.sqrt(np.sum(B**2))

proj_of_A_on_B=(np.dot(A,B)/Magnitude_of_B**2)*B
print('projection of A on B :',proj_of_A_on_B)
```
