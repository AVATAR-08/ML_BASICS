# **MATRIX**

- Vectors Are also Type of  

```python
Matrix_1=np.asarray([[2,3],[4,5]])
print(Matrix_1)
```

Creating Matrices With Random Value

```python
#random.rand generates floating values
Matrix_2=np.random.rand(3,3)
print(Matrix_2)
```

```python
# random.randint generates integer values
Matrix_3=np.random.randint(100, size=(3,3))
print(Matrix_3)
```

Matrix With All Values as 1

```python
# dtype argument is optional
Matrix_4=np.ones((3,3), dtype=int) # By Default dtype is floats
print(Matrix_4)
```

Null Matrix / Zero Matrix

```python

Matrix_5=np.zeros((3,3))
print(Matrix_5)
```

Diagonal Matrix

```python
Matrix_6=np.diag([1,2,3,4])
print(Matrix_6)
```

Identity Matrix

```python
Matrix_7=np.identity(3)
print(Matrix_7)
```

Transpose Of Matrix

```python
# Creating a Matrix of Random Integer Values
Matrix_8=np.random.randint(100, size=(3,4))
print(Matrix_8)
```

```python
# Now creating transpose of Matrix
Transpose_of_Matrix_8=np.transpose(Matrix_8)
print(Transpose_of_Matrix_8)
```

**MATRIX OPERATIONS**


MATRIX ADDITION
- Matrices Should Have The Same Number Of Rows & Column
- Elementwise Addition

```python
A= np.array([[1,2],[3,4]])
B= np.array([[5,6],[7,8]])
C=A+B
print(C)
```

MATRIX SUBTRACTION
- Matrices Should Have The Same Number Of Rows & Column
- Elementwise Subtraction

```python
A= np.array([[1,2],[3,4]])
B= np.array([[5,6],[7,8]])
C=A-B
print(C)
```

MATRIX MULTIPLICATION BY SCALAR

- Elementwise Multiplication

```python
B=np.array([[5,6],[7,8]])
C=2*B
print(C)
```

MULTIPLICATION OF 2 MATRIX

- Number of columns of 1st Matrix Must be Equal To Number of Rows if 2nd Matrix

- The Resultant Matrix will have Same Number Of Rows as The 1st Matrix & The Same Number Of Columns as The 2nd Matrix

```python
A= np.array([[1,2],[3,4]])
B= np.array([[5,6],[7,8]])
C=A*B
print(C)
```
